# YOLO V8 Training


```python
from ultralytics import YOLO


model = YOLO('yolov8x.pt') #Choose the Model Name from above
 
# Training.
results = model.train(
   data='files/custom_data.yaml',
   imgsz=1280,
   epochs=550,
   device="cuda",
   batch=1,
   name='Dotav2Dataset_yolov8x_550'
)

print("Training Done. Find trained models in runs.")
```
